## Merge data

```
data_eucs_transpiration <-
        
        list(
              euc_saligna_residual_conductance_data,
              euc_grandis_residual_conductance_data,
              euc_crebra_residual_conductance_data, 
              euc_dunnii_residual_conductance_data,
              euc_tereticornis_residual_conductance_data,
              euc_melliodora_residual_conductance_data,
              euc_sideroxylum_residual_conductance_data, 
              euc_fibrosa_residual_conductance_data
              ) %>%
            
    reduce(rbind) %>% 
    
    mutate(
           gres = gres_mols_m2_s,
           transpiration_rate = transpiration_double_H2Ommol_per_sec_m2,
           temperature = as.numeric(set_temperature),
           
           # Create Fast vrs Slow groups
           group = factor(ifelse(spcode == "euc_saligna" | spcode == "euc_grandis" | 
                          spcode == "euc_tereticornis" | spcode == "euc_dunnii", 
                          "fast", "slow")),
           
           #above_40 = factor(ifelse(temperature > 40, '>40', "<40")),
           
           # Rename levels in treatment
           treatment = factor(ifelse(vpd_control == "no", "dry_heat", "humid_heat")),
           
           # Convert variablesranspiration_quadratic_mo
           tree_id = factor(tree_id),
           spcode = factor(spcode),
           .keep = 'unused')  %>% 

      
    # Sort columns
    dplyr::select(tree_id, spcode, group, treatment, temperature,  everything()) %>% 
    
    # Remove unused variables
    dplyr::select(-c(branch_dry_matter_content_g_g, sla, branch_sla, lma, 
                     gres))
    
```

## Model fits

### Linear model

```
model_transpiration_net_vpd_effect_linear_0 <-  glmmTMB(
    
                                            # Fixed effects
                                            log(transpiration_rate) ~ treatment * poly(temperature, 1) +
                                                
                                            # Random intercepts and slopes
                                            (1 | spcode) + 
                                                
                                            (1 + treatment | spcode:tree_id),
                                    data = data_eucs_transpiration,
                                    REML = TRUE
                                    )
```

```
VarCorr(model_transpiration_net_vpd_effect_linear_0)
```

```
model_transpiration_net_vpd_effect_linear_1 <-  glmmTMB(
    
                                            # Fixed effects
                                            log(transpiration_rate) ~ treatment * poly(temperature, 1) +
                                                
                                            # Random intercepts and slopes
                                            (1 | spcode) + 
                                                
                                            diag(1 + treatment | spcode:tree_id),
                                    data = data_eucs_transpiration,
                                    REML = TRUE
                                    )
```

```
VarCorr(model_transpiration_net_vpd_effect_linear_1)
```

```
AIC(model_transpiration_net_vpd_effect_linear_0)
```

```
AIC(model_transpiration_net_vpd_effect_linear_1)
```

### Quadratic model

```

model_transpiration_net_vpd_effect_poly_1_0 <- update(model_transpiration_net_vpd_effect_linear_1, 
                                         
                                         # Fixed effects
                                         log(transpiration_rate) ~ treatment * poly(temperature, 2) +
                                         
                                         # Random intercepts and slopes   
                                         (1 | spcode) + 
                                             
                                         (1 + treatment | spcode:tree_id))
```

```
VarCorr(model_transpiration_net_vpd_effect_poly_1_0)
```

```

model_transpiration_net_vpd_effect_poly_1_1 <- update(model_transpiration_net_vpd_effect_linear_1, 
                                         
                                         # Fixed effects
                                         log(transpiration_rate) ~ treatment * poly(temperature, 2) +
                                         
                                         # Random intercepts and slopes   
                                         (1 | spcode) + 
                                             
                                         diag(1 + treatment | spcode:tree_id))
```

```
VarCorr(model_transpiration_net_vpd_effect_poly_1_1)
```

```
AIC(model_transpiration_net_vpd_effect_poly_1_0)
```

```
AIC(model_transpiration_net_vpd_effect_poly_1_1)
```

#### Model comparison linear vrs quadratic fit

```
anova(update(model_transpiration_net_vpd_effect_linear_1, REML = FALSE), 

            update(model_transpiration_net_vpd_effect_poly_1_1, REML = FALSE)) %>% 

    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("Linear", "Quadratic"),
           `Pr(>Chisq)` = ifelse(`Pr(>Chisq)` < 0.001, "<0.001", 
                                 sprintf("%.3f", `Pr(>Chisq)`))) %>%
    
    knitr::kable(digits = 2, caption = "Likelihood-ratio test: linear vs quadratic temperature response")
```

```
# Export table
lrt_table <- anova(update(model_transpiration_net_vpd_effect_linear_1, REML = FALSE),
                   update(model_transpiration_net_vpd_effect_poly_1_1, REML = FALSE)) %>%
    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("Linear", "Quadratic"),
           `Pr(>Chisq)` = ifelse(`Pr(>Chisq)` < 0.001, "<0.001",
                                 sprintf("%.3f", `Pr(>Chisq)`)))

ft <- lrt_table %>%
    flextable() %>%
    colformat_double(digits = 2) %>%
    set_header_labels(Df = "df", Chisq = "\u03C7\u00B2", 
                      `Chi Df` = "\u03C7\u00B2 df",
                      `Pr(>Chisq)` = "P", 
                      deviance = "Deviance") %>%
    
    italic(part = "header", j = "Pr(>Chisq)") %>%
    theme_booktabs() %>%
    font(fontname = "Times New Roman", part = "all") %>%
    fontsize(size = 11, part = "all") %>%
    autofit()


save_as_docx(ft,
             path = "./tables/table_S1_lrt_transpiration_q1.docx",
             pr_section = prop_section(page_size = page_size(orient = "landscape")))
```

### Quadratic model + species linear temperature response with constant-variance

```
model_transpiration_net_vpd_effect_poly_2 <- glmmTMB(
        
                                        # Fixed effects
                                        log(transpiration_rate) ~ treatment * poly(temperature, 2)  + 
    
                                        # Random intercepts and slopes, tree 
                                        # level responses to treatments
                                        (1 + treatment | spcode:tree_id) +
            
                                        # Species-level temperature responses    
                                        (1 + poly(temperature, 1) | spcode),
                                    
                                    data = data_eucs_transpiration,
                                    REML = TRUE)
```

### Quadratic model + species quadratic temperature response with constant-variance

```
model_transpiration_net_vpd_effect_poly_3 <- glmmTMB(
        
                            # Fixed effects
                            log(transpiration_rate) ~ treatment * poly(temperature, 2)  + 
                        
                            # Random intercepts and slopes, tree level 
                            # responses to treatments
                            (1 + treatment | spcode:tree_id) +
                                
                            # Species-level temperature responses    
                            (1 + poly(temperature, 2) | spcode),
                        
                        data = data_eucs_transpiration, 
                        REML = TRUE)
```

```
anova(model_transpiration_net_vpd_effect_poly_1_1,
      model_transpiration_net_vpd_effect_poly_2,
      model_transpiration_net_vpd_effect_poly_3) %>% 

    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("+ Base (species intercepts)",
                   "+ species linear slopes",
                   "+ species curvature"
                   ),
           p = ifelse(is.na(`Pr(>Chisq)`), "",
                      ifelse(`Pr(>Chisq)` < 0.001, "<0.001", sprintf("%.3f", `Pr(>Chisq)`)))) %>%
    select(Model, Df, AIC, logLik, Chisq, `Chi Df`, p) %>%
    knitr::kable(digits = 1,
                 col.names = c("Model", "Df", "AIC", "logLik", "Chisq", "test df", "p"),
                 caption = "Model ladder for residual conductance (LRTs against the preceding model)") %>%
    kableExtra::kable_styling(full_width = FALSE)
```

### Check homogeneity of variances 

```
data_eucs_transpiration$abs_r <- abs(residuals(model_transpiration_net_vpd_effect_poly_3, type = "pearson"))

ggplot(data_eucs_transpiration, aes(factor(temperature), abs_r, 
                                  colour = treatment)) + 
        geom_boxplot() +
        scale_y_continuous(breaks = seq(0, 2, 0.1)) +
        theme_bw() +
        theme(legend.position = "bottom") 
    
```

```
data_eucs_transpiration$r <- residuals(model_transpiration_net_vpd_effect_poly_3, 
                                     type = "pearson")

data_eucs_transpiration %>% 
    group_by(treatment, temperature) %>%
    summarise(sd_r = sd(r), .groups = "drop") %>%
     
    ggplot(aes(temperature, sd_r, colour = treatment))  +
        geom_point(size = 3) + geom_line(linewidth = 1) +
        
        scale_y_continuous(breaks = seq(0, 1, 0.1)) +
        scale_x_continuous(breaks = scales::pretty_breaks(n = 7)) +
        theme_bw() +
        theme(legend.position = "bottom") 
    
```

```
cells <- interaction(data_eucs_transpiration$treatment, 
                     factor(data_eucs_transpiration$temperature))

car::leveneTest(residuals(model_transpiration_net_vpd_effect_poly_3, 
                          type = "pearson"), cells)
```

```
residuals <- residuals(model_transpiration_net_vpd_effect_poly_3, 
                       type = "response")

residuals_standard_deviation <- predict(model_transpiration_net_vpd_effect_poly_3, 
                                        type = "disp")  

# Under homogeneity of variance the ratio should be the same across every cell 
ratio_r_rsd <- residuals / residuals_standard_deviation
```

```
# Test if the ratio is the same across every cell 
car::leveneTest(ratio_r_rsd, cells)
```

### Quadratic model + species quadratic temperature response with variance modeled 

```
model_transpiration_net_vpd_effect_poly_4 <- glmmTMB(
        
                            # Fixed effects
                            log(transpiration_rate) ~ treatment * poly(temperature, 2)  + 
                        
                            # Random intercepts and slopes, tree level 
                            # responses to treatments
                            (1 + treatment | spcode:tree_id) +
                                
                            # Species-level temperature responses    
                            (1 + poly(temperature, 2) | spcode),
                            
                            # No constant variance 
                            dispformula = ~ treatment * poly(temperature, 2),
                            
                        data = data_eucs_transpiration, 
                        REML = TRUE)
```

```
residuals <- residuals(model_transpiration_net_vpd_effect_poly_4, 
                       type = "response")

residuals_standard_deviation <- predict(model_transpiration_net_vpd_effect_poly_4, 
                                        type = "disp")  

ratio_r_rsd <- residuals / residuals_standard_deviation
```

```
car::leveneTest(ratio_r_rsd, cells)
```

## Model Fits

### Linear model

```
model_transpiration_group_responses_linear <- glmmTMB(
        
            # Fixed effects
            log(transpiration_rate) ~ treatment * poly(temperature, 1) * group + 
    
            # Random intercepts and slopes, tree level responses to treatments
            #(1 | spcode) + (1 + treatment | spcode:tree_id),
            (1 | spcode) + (1 | spcode:tree_id),
  
    data = data_eucs_transpiration, 
    REML = TRUE)
```

### Quadratic model

```
model_transpiration_group_responses_poly_1 <- glmmTMB(
        
        # Fixed effects
        log(transpiration_rate) ~ treatment * poly(temperature, 2) * group + 
    
            # Random intercepts and slopes, tree level responses to treatments
            (1 | spcode) + (1 | spcode:tree_id),
  
    data = data_eucs_transpiration, 
    REML = TRUE)
```

#### Model comparison linear vrs quadratic fit


```
anova(update(model_transpiration_group_responses_linear, REML = FALSE), 
      update(model_transpiration_group_responses_poly_1, REML = FALSE)) %>% 
    
    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("Linear", "Quadratic"),
           `Pr(>Chisq)` = ifelse(`Pr(>Chisq)` < 0.001, "<0.001", 
                                 sprintf("%.3f", `Pr(>Chisq)`))) %>%
    
    knitr::kable(digits = 2, caption = "Likelihood-ratio test: linear vs quadratic temperature response")
```

```

lrt_table <- anova(update(model_transpiration_group_responses_linear, REML = FALSE), 
      update(model_transpiration_group_responses_poly_1, REML = FALSE)) %>%
    
    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("Linear", "Quadratic"),
           `Pr(>Chisq)` = ifelse(`Pr(>Chisq)` < 0.001, "<0.001",
                                 sprintf("%.3f", `Pr(>Chisq)`)))

ft <- lrt_table %>%
    flextable() %>%
    colformat_double(digits = 2) %>%
    set_header_labels(Df = "df", Chisq = "\u03C7\u00B2", 
                      `Chi Df` = "\u03C7\u00B2 df",
                      `Pr(>Chisq)` = "P", 
                      deviance = "Deviance") %>%
    
    italic(part = "header", j = "Pr(>Chisq)") %>%
    theme_booktabs() %>%
    font(fontname = "Times New Roman", part = "all") %>%
    fontsize(size = 11, part = "all") %>%
    autofit()


save_as_docx(ft,
             path = "./tables/table_S1_lrt_transpiration_q2.docx",
             pr_section = prop_section(page_size = page_size(orient = "landscape")))
```

### Quadratic model + species linear temperature response with constant-variance

```
model_transpiration_group_responses_poly_2 <- glmmTMB(
        
        # Fixed effects
        log(transpiration_rate) ~ treatment * poly(temperature, 2) * group + 
    
            # Random intercepts and slopes, tree level responses to treatments
             (1 + treatment | spcode:tree_id) +
         
            # Species-level temperature responses 
            (1 + poly(temperature, 1) | spcode), 
  
    data = data_eucs_transpiration, 
    REML = TRUE
    )
```

### Quadratic model + species quadratic temperature response with constant-variance  

```
model_transpiration_group_responses_poly_3 <- glmmTMB(
        
        # Fixed effects
        log(transpiration_rate) ~ treatment * poly(temperature, 2) * group + 
    
            # Random intercepts and slopes, tree level responses to treatments
            (1 + treatment | spcode:tree_id) +
         
            # Species-level temperature responses 
            diag(1 + poly(temperature, 2) | spcode), 
  
    data = data_eucs_transpiration, 
    REML = TRUE)
```

### Model comparison 

```
anova( 
      model_transpiration_group_responses_poly_1,
      model_transpiration_group_responses_poly_2,
      model_transpiration_group_responses_poly_3)  %>%
    
    as.data.frame() %>%
    tibble::rownames_to_column("Model") %>%
    mutate(Model = c("+ Base (species intercepts)",
                   "+ species linear slopes",
                   "+ species curvature"),
           p = ifelse(is.na(`Pr(>Chisq)`), "",
                      ifelse(`Pr(>Chisq)` < 0.001, "<0.001", sprintf("%.3f", `Pr(>Chisq)`)))) %>%
    select(Model, Df, AIC, logLik, Chisq, `Chi Df`, p) %>%
    knitr::kable(digits = 1,
                 col.names = c("Model", "Df", "AIC", "logLik", "Chisq", "test df", "p"),
                 caption = "Model ladder for residual conductance (LRTs against the preceding model)") %>%
    kableExtra::kable_styling(full_width = FALSE)
```

### Check homogeneity of variances 

```
data_eucs_transpiration$abs_r <- abs(residuals(model_transpiration_group_responses_poly_2, 
                                      type = "pearson"))

ggplot(data_eucs_transpiration, aes(factor(temperature), abs_r, colour = treatment)) +
    geom_boxplot() +
        scale_y_continuous(breaks = seq(0, 2, 0.1)) +
        theme_bw() +
        theme(legend.position = "bottom") 
```

```
data_eucs_transpiration$r <- residuals(model_transpiration_group_responses_poly_2, 
                              type = "pearson")

data_eucs_transpiration %>% 
    group_by(treatment, temperature) %>%
    summarise(sd_r = sd(r), .groups = "drop") %>%
     
    ggplot(aes(temperature, sd_r, colour = treatment))  +
        geom_point(size = 3) + geom_line(linewidth = 1) +
        
        scale_y_continuous(breaks = seq(0, 1, 0.1)) +
        scale_x_continuous(breaks = scales::pretty_breaks(n = 7)) +
        theme_bw() +
        theme(legend.position = "bottom") 
```

```
cells <- interaction(data_eucs_transpiration$treatment, 
                     factor(data_eucs_transpiration$temperature))

car::leveneTest(residuals(model_transpiration_group_responses_poly_2, 
                          type = "pearson"), 
                cells)
```

```
residuals <- residuals(model_transpiration_group_responses_poly_2, type = "response")

residuals_standard_deviation <- predict(model_transpiration_group_responses_poly_2, 
                                        type = "disp")  

# Under homogeneity of variace the ratio should be the same across every cell 
ratio_r_rsd <- residuals / residuals_standard_deviation
```

```
# Test if the ratio is the same across every cell 
car::leveneTest(ratio_r_rsd, cells)
```

### Quadratic model + species quadratic temperature response with variance modeled 

```

model_transpiration_group_responses_poly_4 <- glmmTMB(
        
                            # Fixed effects
                            log(transpiration_rate) ~ treatment * poly(temperature, 2) * group + 
                        
                            # Random intercepts and slopes, tree level 
                            # responses to treatments
                            (1 + treatment | spcode:tree_id) +
                                
                            # Species-level temperature responses    
                            (1 + poly(temperature, 2) | spcode),
                            
                            # No constant variance 
                            dispformula = ~ treatment * poly(temperature, 2),
                            
                        data = data_eucs_transpiration, 
                        REML = TRUE)
```

```
residuals <- residuals(model_transpiration_group_responses_poly_4, type = "response")

residuals_standard_deviation <- predict(model_transpiration_group_responses_poly_4, type = "disp")  

ratio_r_rsd <- residuals / residuals_standard_deviation
```
