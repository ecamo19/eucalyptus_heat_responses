# 1. Load data 

```
data_vpd_at_hfe <- read.csv('./03_daily_vpd_hfe.csv', header = T) 
```

```
data_vpd_at_hfe_cleaned <- 
    data_vpd_at_hfe %>% 
        mutate(date = ymd(Date), .keep = "unused") %>%
        clean_names() 
```

```
data_vpd_at_hfe_cleaned %>% 
    mutate(year = year(date)) %>%
    summarise(mean_vpd = mean(vpd_max),
              sd_vpd   = sd(vpd_max),
              )
```
