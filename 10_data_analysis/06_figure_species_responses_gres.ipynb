## Merge data

```
data_eucs <- 
                list(
                    euc_saligna_residual_conductance_data,
                    euc_grandis_residual_conductance_data,
                    euc_crebra_residual_conductance_data,
                    euc_dunnii_residual_conductance_data,
                    euc_tereticornis_residual_conductance_data,
                    euc_melliodora_residual_conductance_data,
                    euc_sideroxylum_residual_conductance_data,
                    euc_fibrosa_residual_conductance_data) %>%
            
    reduce(rbind) %>% 
    
    mutate(
           # Log transform data for reducing variability
           gres = gres_mols_m2_s,
           dehydration_rate = transpiration_double_H2Ommol_per_sec_m2,
           temperature = as.numeric(set_temperature),
           
           # Rename levels in treatment
           treatment = factor(if_else(vpd_control == "no", "dry_heat", "humid_heat")),
           
           # Convert variables
           tree_id = factor(tree_id),
           spcode = factor(spcode),
           .keep = 'unused') %>% 
    
    # Sort columns
    dplyr::select(tree_id, spcode, treatment, temperature,  gres) 
    
```

```
data_eucs_summary_stats_gres <- 
    data_eucs %>% 
    select(spcode, temperature, treatment, gres) %>% 
    group_by(spcode, temperature, treatment) %>% 
    summarise(median_gres = median(gres),
              sd_val   = sd(gres, na.rm = TRUE),
              n        = n(),
              se       = sd_val / sqrt(n)) %>% 
    filter(!(temperature == 25 & treatment == "humid_heat"))  %>%   
    mutate(treatment = factor(if_else(temperature == 25, 
                               "baseline", as.character(treatment)))) 
```

# Residual conductance

```
(supplemental_figure_3_gres <- 
    ggplot(data = data_eucs_summary_stats_gres, 
          aes(temperature, median_gres, colour = treatment)) +
    
            # Raw data at 25C
            geom_point(data = filter(data_eucs, temperature == 25 ),
                       
                       aes(temperature, gres, group = temperature, 
                              fill = NULL),
                       alpha = 0.30,
                       inherit.aes = FALSE,
                       width = 1, 
                       shape = 18,
                       show.legend = FALSE,
                       size = 8,
                       outlier.shape = NA,
                       colour = "#6E6E6E",
                       linewidth = 1.5,
                       position = position_jitterdodge(jitter.width  = 2.5,
                                           jitter.height = 0,
                                           dodge.width   = 3.5,
                                           seed = 1)) +
    
            # Raw data above 25C
            geom_point(data = filter(data_eucs, temperature > 25),
                       
                       aes(temperature, gres, colour = treatment,
                           fill = NULL, group = interaction(temperature, 
                                                            treatment)),
                       
                       alpha = 0.15,
                       shape = 18,
                       show.legend = FALSE,
                       size = 8,
                       width = 5, 
                       outlier.shape = NA,
                       linewidth = 1.5,
                       position = position_jitterdodge(jitter.width  = 2.5,
                                           jitter.height = 0,
                                           dodge.width   = 3.5,
                                           seed = 1)) +
                   
            facet_wrap(~ factor(spcode, levels = c("euc_fibrosa","euc_melliodora",
                                                   "euc_sideroxylum","euc_crebra",
                
                                                    "euc_dunnii", "euc_grandis",
                                                   "euc_saligna", "euc_tereticornis"),
                                
                       labels = c("Eucalyptus fibrosa", "Eucalyptus melliodora", 
                                  "Eucalyptus sideroxylom", "Eucalyptus crebra",
                                  
                                  "Eucalyptus dunnii", "Eucalyptus grandis", 
                                  "Eucalyptus saligna", "Eucalyptus tereticornis")),
                       ncol = 4) +
            
            # Add SE for 25C
            geom_errorbar(data = filter(data_eucs_summary_stats_gres, 
                                    temperature == 25),
                      
                      aes(ymin = median_gres - se, 
                          ymax =  median_gres + se, 
                          group = treatment),
                      
                      position = position_dodge(width = 3.5),
                      show.legend = FALSE,
                      colour = "black",
                      width = 2.5,
                      linewidth = 0.75) +
            
            # Add point estimates for 25C
            geom_point(data = filter(data_eucs_summary_stats_gres, 
                                 (temperature == 25)),
                       aes(fill = treatment),
                       shape = 23,
                       size = 5,
                       stroke = 0.5,
                       colour = "black",
                       position = position_dodge(width = 3.5)) +
    
            # Add point estimates everything else
            geom_errorbar(data = filter(data_eucs_summary_stats_gres, 
                                    temperature > 25),
                      
                      aes(ymin = median_gres - se, 
                          ymax =  median_gres + se, 
                          group = treatment),
                      
                      position = position_dodge(width = 3.5),
                      show.legend = FALSE,
                      colour = "black",
                      width = 4.5,
                      linewidth = 0.75) +
            
            geom_point(data = filter(data_eucs_summary_stats_gres, 
                                 (temperature > 25)),
                       aes(fill = treatment),
                       shape = 23,
                       size = 5,
                       stroke = 0.5,
                       colour = "black",
                       position = position_dodge(width = 3.5)) +
            
            # Theme
            labs(x = "Temperature (°C)", y = expression(g[res]~(mol~m^-2~s^-1))) +
      
            theme_bw(base_size = 45) +
        
           theme(legend.position = "inside",
              legend.title = element_blank(),
              legend.justification = c(0.005, 0.99),
              legend.position.inside = c(0.005, 0.99),
              legend.text = element_text(size = 30),
              legend.background = element_blank(),
              panel.grid.major = element_blank(),
              panel.grid.minor = element_blank(),
              strip.background = element_rect(fill = "white",colour = NA),
              strip.text = element_text(colour = "black",face = "italic"),
              panel.border = element_rect(colour = "black",fill = NA,
                                          linewidth = 4),
              axis.text    = element_text(colour = "black"),
              axis.ticks   = element_line(colour = "black")) +
            
            # Set colors
            scale_fill_manual(values = c("baseline" = "#6E6E6E",
                                         "humid_heat" = "#293352",
                                         "dry_heat" = "#9e1523"),
                              
                              labels = c("baseline" = "Baseline",
                                         "humid_heat" = "Humid heat",
                                         "dry_heat" = "Dry heat"),
                              
                              breaks = c("baseline", "humid_heat", "dry_heat")) + 
        
            scale_color_manual(values = c("baseline" = "#6E6E6E" ,
                                          "humid_heat" ="#293352",
                                          "dry_heat" = "#9e1523")) +
         
            scale_y_continuous(breaks = seq(0, 0.05, 0.01)) +
                               
            
            coord_cartesian(ylim = c(0, 0.06))+ 
        
            scale_x_continuous(breaks = scales::pretty_breaks(n = 7)) +
            guides(fill = guide_legend(override.aes = list(shape = 23, size = 8, stroke = 1.5)),colour = "none")
)       
```
